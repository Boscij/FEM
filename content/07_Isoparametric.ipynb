# 08 – Isoparametrische Vierecks-Elemente

Mit **bilinearen 4-Knoten Elementen** berechnen wir hier die Verformung eines 2D-Bauteils
im **ebenen Spannungszustand** (engl. plane stress, dünne Platte, $\sigma_{zz}=0$).

## Was ist neu gegenüber Stab, Balken und Membran?

| | Stab/Balken | Membran  | Isoparametrisch (4 Knoten) |
|---|---|---|---|
| Knoten pro Element | 2 | 3 | **4** |
| Freiheitsgrade pro Knoten | 2 oder 3 | 1 ($w$) | **2** ($u, v$) |
| Lokale ESM | $4\!\times\!4$ bzw. $6\!\times\!6$ | $3\!\times\!3$ | $8 \times 8$ |
| Dehnung im Element | konstant | konstant | **bilinear** |
| Integration | analytisch | analytisch | **numerisch (Gauss)** |

**Isoparametrisch** bedeutet, dass dieselben Formfunktionen $N_i$ sowohl die Geometrie als auch
das Verschiebungsfeld interpolieren

$$
x(\xi,\eta) = \sum_{i=1}^{4} N_i(\xi,\eta)\,x_i, \qquad
u(\xi,\eta) = \sum_{i=1}^{4} N_i(\xi,\eta)\,u_i .
$$

## Referenzelement und Formfunktionen

Wir rechnen nicht direkt im $(x,y)$-Raum, sondern auf einem **Einheitsquadrat**
$[-1,+1] \times [-1,+1]$ mit Koordinaten $(\xi, \eta)$.

![Referenzquadrat mit Gauss-Punkten](img/isoparametric_refsquare.png)

Die **bilinearen Formfunktionen** sind

$$
\begin{aligned}
N_1(\xi,\eta) &= \tfrac{1}{4}(1-\xi)(1-\eta) \\
N_2(\xi,\eta) &= \tfrac{1}{4}(1+\xi)(1-\eta) \\
N_3(\xi,\eta) &= \tfrac{1}{4}(1+\xi)(1+\eta) \\
N_4(\xi,\eta) &= \tfrac{1}{4}(1-\xi)(1+\eta)
\end{aligned}
$$

Jede Funktion ist $1$ am eigenen Knoten und $0$ an allen drei anderen.

## Jacobi-Matrix und B-Matrix

Die Dehnung braucht die Ableitungen $\partial N_i/\partial x$ und $\partial N_i/\partial y$.
Im Referenzraum kennen wir aber nur $\partial N_i/\partial \xi$ und $\partial N_i/\partial \eta$.
Die **Jacobi-Matrix** $\underline{\underline{J}}$ vermittelt zwischen beiden:

$$
\underline{\underline{J}} =
\begin{pmatrix}
 \partial x/\partial \xi & \partial y/\partial \xi \\
 \partial x/\partial \eta & \partial y/\partial \eta
\end{pmatrix},
\qquad
\begin{pmatrix} \partial N_i/\partial x \\ \partial N_i/\partial y \end{pmatrix}
=
\underline{\underline{J}}^{-1}
\begin{pmatrix} \partial N_i/\partial \xi \\ \partial N_i/\partial \eta \end{pmatrix} .
$$

Die **Determinante** $\det(\underline{\underline{J}})$ ist der Flächen-Skalierungsfaktor

$$
\mathrm{d}x\,\mathrm{d}y = \det(\underline{\underline{J}})\,\mathrm{d}\xi\,\mathrm{d}\eta .
$$

Die **B-Matrix** ($3 \times 8$) verbindet Knotenverschiebungen $\boldsymbol{u}^e = [u_1, v_1, \dots, u_4, v_4]^\top$
mit den Dehnungen $\boldsymbol{\varepsilon} = [\varepsilon_{xx}, \varepsilon_{yy}, \gamma_{xy}]^\top$:

$$
\underline{\underline{B}} =
\begin{pmatrix}
 N_{1,x} & 0      & N_{2,x} & 0      & N_{3,x} & 0      & N_{4,x} & 0      \\
 0       & N_{1,y}& 0       & N_{2,y}& 0       & N_{3,y}& 0       & N_{4,y}\\
 N_{1,y} & N_{1,x}& N_{2,y} & N_{2,x}& N_{3,y} & N_{3,x}& N_{4,y} & N_{4,x}
\end{pmatrix} .
$$

## Materialgesetz: Elastizitätsmatrix $\underline{\underline{E}}$

Für isotropes, linear-elastisches Material gilt $\boldsymbol{\sigma} = \underline{\underline{E}}\,\boldsymbol{\varepsilon}$. In der Ebene gibt es zwei Varianten:

**Ebener Spannungszustand (ESZ)** -- dünne Scheibe, $\sigma_{zz} = 0$:

$$
\underline{\underline{E}}_\text{ESZ} = \frac{E}{1-\nu^2}
\begin{pmatrix}
 1 & \nu & 0 \\
 \nu & 1 & 0 \\
 0 & 0 & (1-\nu)/2
\end{pmatrix} .
$$

**Ebener Verzerrungszustand (EVZ)** -- langer Körper, $\varepsilon_{zz} = 0$:

$$
\underline{\underline{E}}_\text{EVZ} = \frac{E}{(1+\nu)(1-2\nu)}
\begin{pmatrix}
 1-\nu & \nu & 0 \\
 \nu & 1-\nu & 0 \\
 0 & 0 & (1-2\nu)/2
\end{pmatrix} .
$$

$E$ ist der E-Modul, $\nu$ die Querkontraktionszahl (Poisson). Welcher Zustand zutrifft, hängt von der Geometrie ab -- ein dünnes Blech rechnet man als ESZ, einen langen Damm-Querschnitt als EVZ.

## Gauss-Integration

Die Element-Steifigkeitsmatrix ergibt sich aus dem Integral

$$
\underline{\underline{K}}^e = t \int_{-1}^{+1}\!\int_{-1}^{+1}
\underline{\underline{B}}^\top \underline{\underline{E}}\,\underline{\underline{B}}\;
\det(\underline{\underline{J}})\;\mathrm{d}\xi\,\mathrm{d}\eta ,
$$

wobei $t$ die Dicke ist. Da $\underline{\underline{B}}$ und $\det(\underline{\underline{J}})$ von
$(\xi,\eta)$ abhängen, integrieren wir **numerisch** per **2×2 Gauss-Quadratur**:

$$
\xi_p, \eta_q \in \Bigl\{-\tfrac{1}{\sqrt{3}},\; +\tfrac{1}{\sqrt{3}}\Bigr\}, \qquad
\omega_p = \omega_q = 1, \qquad
\underline{\underline{K}}^e \approx t \sum_{p,q} \omega_p \omega_q\;
\underline{\underline{B}}^\top \underline{\underline{E}}\,\underline{\underline{B}}\;
\det(\underline{\underline{J}})\Big|_{(\xi_p, \eta_q)} .
$$

## Modelldefinition: Kragarm aus Stahlblech

- Länge $L = 100\,\mathrm{mm}$, Höhe $h = 20\,\mathrm{mm}$, Dicke $t = 1\,\mathrm{mm}$
- Material Stahl, $E = 210\,\mathrm{GPa}$, $\nu = 0.3$
- Linker Rand komplett eingespannt, am rechten Rand wirkt eine vertikale Streckenlast $F = -1000\,\mathrm{N}$ nach unten

Wir verwenden ein **regelmässiges iso.-4-Knoten-Netz**. Knoten-Reihenfolge je Element entgegen Uhrzeigersinn:
$N_1, N_2, N_3, N_4 = $ SW, SE, NE, NW.

```python
# Geometrie und Material (alles in SI - m, Pa, N)
L  = 0.1          # Laenge [m]
h  = 0.02         # Hoehe  [m]
t  = 1e-3         # Dicke  [m]
E  = 210e9        # E-Modul [Pa]
nu = 0.3          # Poisson

# Regelmaessiges iso.-4-Knoten-Netz
nx, ny = 25, 5    # Anzahl Elemente in x bzw. y
x = np.linspace(0, L, nx + 1)
y = np.linspace(0, h, ny + 1)
X, Y = np.meshgrid(x, y)
nodes = np.column_stack((X.ravel(), Y.ravel()))   # (n_knoten, 2)

# Vierecks-Elemente (Knoten in Reihenfolge SW, SE, NE, NW - entgegen Uhrzeigersinn)
elements = []
for i in range(ny):
    for j in range(nx):
        sw = i*(nx+1) + j
        se = sw + 1
        nw = sw + (nx+1)
        ne = nw + 1
        elements.append([sw, se, ne, nw])
elements = np.array(elements)

print(f"Knoten:   {len(nodes)}")
print(f"Elemente: {len(elements)}  ({nx} x {ny})")
```

## Randbedingungen und Lasten als boolesche Masken

- `fixed`:  boolesche Maske der Länge $2 N_{\text{Knoten}}$, `True` an gesperrten DOFs.
- `F`:      globaler Lastvektor der gleichen Länge.
- Reihenfolge der DOFs:  $[u_{x,1}, u_{y,1},\; u_{x,2}, u_{y,2},\; \dots]$
  also Knoten $k$ hat DOFs $2k$ (x) und $2k+1$ (y).

```python
tol = 1e-9
n_dof = 2 * len(nodes)

# Einspannung am linken Rand:  u_x = u_y = 0
left   = np.where(nodes[:, 0] < tol)[0]
fixed  = np.zeros(n_dof, dtype=bool)
fixed[2*left    ] = True   # u_x
fixed[2*left + 1] = True   # u_y

# Vertikale Streckenlast am rechten Rand (gleichmaessig auf die Randknoten verteilt)
F_total = -1000.0
right   = np.where(nodes[:, 0] > L - tol)[0]
F       = np.zeros(n_dof)
F[2*right + 1] = F_total / len(right)   # nur u_y-Komponente belastet

# Skizze
fig, ax = plt.subplots(figsize=(12, 3))
ax.set_aspect("equal")
for el in elements:
    idx = list(el) + [el[0]]
    ax.plot(nodes[idx, 0], nodes[idx, 1], "k-", linewidth=0.4)
ax.plot(nodes[left,  0], nodes[left,  1], "g>", markersize=10, label="Einspannung")
for k in right:
    ax.annotate("", xy=(nodes[k, 0], nodes[k, 1] - 0.005),
                xytext=(nodes[k, 0], nodes[k, 1]),
                arrowprops=dict(arrowstyle="->", color="r", linewidth=1.4))
ax.set_xlabel("x [m]"); ax.set_ylabel("y [m]")
ax.set_title(f"Kragarm: {nx}×{ny} iso.-4-Knoten-Elemente, Einspannung links (grün), Last rechts (rot)")
ax.legend(loc="upper left")
ax.margins(0.1)
plt.show()
```

## B-Matrix am Punkt $(\xi, \eta)$

Diese Funktion berechnet $\underline{\underline{B}}$ und $\det(\underline{\underline{J}})$ an einem
beliebigen Punkt im Element. Wir nutzen sie sowohl in der Gauss-Integration als auch im Postprocessing.

```python
def B_matrix(xy_e, xi, eta):
    """B (3x8) und det(J) am Punkt (xi, eta) im iso.-4-Knoten-Element xy_e (4 x 2)."""

    # Ableitungen der vier Formfunktionen N_i nach xi und eta
    dN_dxi  = np.array([-(1-eta), +(1-eta), +(1+eta), -(1+eta)]) / 4
    dN_deta = np.array([-(1-xi ), -(1+xi ), +(1+xi ), +(1-xi )]) / 4

    # Jacobi-Matrix J (2x2)  und  Inverse
    J = np.array([[dN_dxi  @ xy_e[:, 0], dN_dxi  @ xy_e[:, 1]],
                  [dN_deta @ xy_e[:, 0], dN_deta @ xy_e[:, 1]]])
    det_J = np.linalg.det(J)
    inv_J = np.linalg.inv(J)

    # Ableitungen im physikalischen Raum
    dN_dx = inv_J[0, 0]*dN_dxi + inv_J[0, 1]*dN_deta
    dN_dy = inv_J[1, 0]*dN_dxi + inv_J[1, 1]*dN_deta

    # B-Matrix (3 x 8)
    B = np.zeros((3, 8))
    B[0, 0::2] = dN_dx   # epsilon_xx = sum N_i,x u_i
    B[1, 1::2] = dN_dy   # epsilon_yy = sum N_i,y v_i
    B[2, 0::2] = dN_dy   # gamma_xy
    B[2, 1::2] = dN_dx
    return B, det_J
```

## Elastizitätsmatrix

```python
def E_matrix(E, nu, state="ESZ"):
    """Elastizitaetsmatrix E (3x3) fuer ebenen Spannungs- (ESZ) oder Verzerrungszustand (EVZ)."""
    if state == "ESZ":
        return E / (1 - nu**2) * np.array([[1,  nu,        0],
                                            [nu, 1,         0],
                                            [0,  0, (1-nu)/2]])
    elif state == "EVZ":
        return E / ((1 + nu) * (1 - 2*nu)) * np.array(
            [[1 - nu, nu,            0],
             [nu,     1 - nu,        0],
             [0,      0,      (1 - 2*nu)/2]])
    else:
        raise ValueError("state muss 'ESZ' oder 'EVZ' sein")


state = "ESZ"          # Wechsle auf "EVZ" fuer ebenen Verzerrungszustand
E_mat = E_matrix(E, nu, state)
print(f"E [Pa] ({state}):\n{E_mat}")
```

## Element-Steifigkeitsmatrix per 2×2 Gauss-Quadratur

```python
# Gauss-Punkte und Gewichte fuer 2x2-Integration
gp = np.array([-1/np.sqrt(3), +1/np.sqrt(3)])
gw = np.array([1.0, 1.0])


def element_stiffness(xy_e, E_mat, t):
    """Element-Steifigkeitsmatrix K_e (8x8) via 2x2-Gauss-Integration."""
    Ke = np.zeros((8, 8))
    for p in range(2):
        for q in range(2):
            B, det_J = B_matrix(xy_e, gp[p], gp[q])
            Ke += B.T @ E_mat @ B * det_J * gw[p] * gw[q] * t
    return Ke


# Test: 1 mm x 1 mm Quadrat-Element
xy_test = np.array([[0, 0], [1e-3, 0], [1e-3, 1e-3], [0, 1e-3]])
Ke_test = element_stiffness(xy_test, E_mat, t)
print(f"K_e Test-Element shape  = {Ke_test.shape}")
print(f"Symmetrie    max|K_e - K_e^T| = {np.max(np.abs(Ke_test - Ke_test.T)):.2e}")
print(f"Rang von K_e = {np.linalg.matrix_rank(Ke_test)}  (=5, mit 3 Starrkoerpermoden)")
```

## Assemblierung

Jedes Element hat 8 DOFs (2 pro Knoten, 4 Knoten). Für ein Element mit Knoten $n_0, n_1, n_2, n_3$
lauten die globalen DOF-Indizes

$$
[\,2n_0,\, 2n_0{+}1,\; 2n_1,\, 2n_1{+}1,\; 2n_2,\, 2n_2{+}1,\; 2n_3,\, 2n_3{+}1\,] .
$$

Im Code: `dofs = np.array([2*n+d for n in el for d in (0, 1)])`.

```python
def assemble(nodes, elements, E_mat, t):
    """Globale Steifigkeitsmatrix K (n_dof x n_dof) durch Aufsummieren der K_e."""
    n_dof = 2 * len(nodes)
    K = np.zeros((n_dof, n_dof))
    for el in elements:
        dofs = np.array([2*n + d for n in el for d in (0, 1)])
        Ke   = element_stiffness(nodes[el], E_mat, t)
        K[np.ix_(dofs, dofs)] += Ke
    return K


K = assemble(nodes, elements, E_mat, t)
print(f"K Dimension: {K.shape}")
print(f"Rang von K:  {np.linalg.matrix_rank(K)} von {K.shape[0]}  (3 Starrkoerpermoden)")
print(f"Symmetrie    max|K - K^T| = {np.max(np.abs(K - K.T)):.2e}")
```

## Lösen mit Randbedingungen

Wie bei allen vorigen Elementen: nur das reduzierte System der freien DOFs lösen.

```python
def solve_system(K, F, fixed):
    """Loest K U = F mit Dirichlet RB U=0 auf 'fixed'. Liefert U und Reaktionskraefte."""
    free   = ~fixed
    U      = np.zeros_like(F)
    U[free] = np.linalg.solve(K[np.ix_(free, free)], F[free])
    R       = K @ U                     # interne Knotenkraefte
    return U, R


U, R = solve_system(K, F, fixed)

ux, uy = U[0::2], U[1::2]
u_max  = np.max(np.sqrt(ux**2 + uy**2))
w_tip  = uy[np.argmax(nodes[:,0])]      # vertikale Auslenkung am rechten Rand

print(f"Max Verschiebung |u|      = {u_max*1e3:.4f} mm")
print(f"Spitzen-Auslenkung w_tip  = {w_tip*1e3:.4f} mm")

# Vergleich mit Bernoulli-Balken: w = F L^3 / (3 E I_z)
I_z    = t * h**3 / 12
w_beam = abs(F_total) * L**3 / (3 * E * I_z)
print(f"\nBernoulli-Balken (ohne Schub):  w = F L^3 / (3 E I_z) = {w_beam*1e3:.4f} mm")
```

## Postprocessing: Dehnung und Spannung

Aus Knotenverschiebungen und B-Matrix folgen die Dehnungen, daraus per Hooke die Spannungen

$$
\boldsymbol{\varepsilon} = \underline{\underline{B}}\,\boldsymbol{u}^e,
\qquad
\boldsymbol{\sigma} = \underline{\underline{E}}\,\boldsymbol{\varepsilon}.
$$

Da jeder Knoten zu mehreren Elementen gehört, mitteln wir die Spannungsbeiträge je Knoten.
Die **Von-Mises-Vergleichsspannung** im 2D-Fall:

$$
\sigma_{\mathrm{vM}} = \sqrt{\sigma_{xx}^2 - \sigma_{xx}\sigma_{yy} + \sigma_{yy}^2 + 3\sigma_{xy}^2} .
$$

```python
# Spannungen an den vier Knoten jedes Elements auswerten und mitteln
node_corners = [(-1,-1), (+1,-1), (+1,+1), (-1,+1)]
sig    = np.zeros((len(nodes), 3))       # spaeter mittlere Spannung je Knoten
count  = np.zeros(len(nodes), dtype=int)

for el in elements:
    xy_e = nodes[el]
    dofs = np.array([2*n + d for n in el for d in (0, 1)])
    u_e  = U[dofs]
    for k, (xi_n, eta_n) in enumerate(node_corners):
        B, _ = B_matrix(xy_e, xi_n, eta_n)
        sig[el[k]] += E_mat @ B @ u_e
        count[el[k]] += 1
sig /= count[:, None]

# Hauptkomponenten + Von-Mises
S11, S22, S12 = sig.T
SVM = np.sqrt(S11**2 - S11*S22 + S22**2 + 3*S12**2)

print(f"Max Von-Mises-Spannung = {SVM.max()/1e6:.2f} MPa  (am eingespannten Rand)")
print(f"Max |sigma_xx|         = {np.abs(S11).max()/1e6:.2f} MPa")
```

## Visualisierung: verformtes Netz mit Von-Mises-Spannung

```python
magnify = 5     # Verstaerkungsfaktor fuer die Sichtbarkeit

# Verformte Knotenkoordinaten
x_def = nodes[:, 0] + magnify * ux
y_def = nodes[:, 1] + magnify * uy

# Quad-Elemente in 2 Dreiecke pro Quad zerlegen (fuer tricontourf)
tris = np.vstack((elements[:, [0, 1, 2]], elements[:, [0, 2, 3]]))

fig, ax = plt.subplots(figsize=(13, 4))
ax.set_aspect("equal")

# Unverformtes Netz hellgrau im Hintergrund
for el in elements:
    idx = list(el) + [el[0]]
    ax.plot(nodes[idx, 0], nodes[idx, 1], color="0.75", linewidth=0.4, linestyle="--")

# Spannungs-Konturplot auf dem verformten Netz
contour = ax.tricontourf(x_def, y_def, tris, SVM/1e6, levels=12, cmap="viridis_r")
plt.colorbar(contour, ax=ax, label="Von-Mises [MPa]", shrink=0.85)

# Verformtes Netz darueber
for el in elements:
    idx = list(el) + [el[0]]
    ax.plot(x_def[idx], y_def[idx], "k-", linewidth=0.4)

ax.set_xlabel("x [m]"); ax.set_ylabel("y [m]")
ax.set_title(rf"Kragarm, Verformung ×{magnify}, "
             rf"$F = {F_total:.0f}$ N, $w_{{\mathrm{{tip}}}} = {w_tip*1e3:.3f}$ mm")
plt.show()
```

## Was, wenn?

- **Feineres Netz** (`nx`, `ny` grösser)  -- konvergieren Verschiebung und Spannung?
  Vergleiche mit der Bernoulli-Lösung oben.
- **Höheres Verhältnis** $L/h$ (schlankerer Balken)  -- wird die FEM-Lösung näher zu Bernoulli?
- **iso.-4-Knoten in Biegung neigt zu "shear locking"**  -- bei sehr schlanken Balken werden iso.-4-Knoten-Elemente
  künstlich steif. Abhilfe: höherwertige Elemente (Q8/Q9) oder reduzierte Integration.
